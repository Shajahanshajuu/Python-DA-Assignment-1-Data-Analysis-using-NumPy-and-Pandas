```python
import numpy as np
```

```python
# Step 1 : Create 1D array for week 1 temparatures
temperatures_w1 = np.array([22.5, 25.3, 20.8, 23.4, 26.1, 24.8, 21.9])
print("temperatures_w1:", temperatures_w1)
```

```python
# Step 2 : Inspection and properties
print("Shape of temperatures_w1:", temperatures_w1.shape)
print("Data type of temperatures_w1:", temperatures_w1.dtype)
print("Number of elements in temperatures_w1:", temperatures_w1.size)
```

```python
# Step 3 : Array operations (Celsius to Fahrenheit)
fahrenheit_w1 = (temperatures_w1 * 9/5) + 32
print("fahrenheit_w1:", fahrenheit_w1)

# Statistical computations
max_temp = np.max(temperatures_w1)
min_temp = np.min(temperatures_w1)
mean_temp = np.mean(temperatures_w1)

print("Maximum Temperature (Celsius):", max_temp)
print("Minimum Temperature (Celsius):", min_temp)
print("Mean Temperature (Celsius):", mean_temp)
```

```python
# Step 4 : Array slicing and Indexing
First_three_days = temperatures_w1[:3]
weekend = temperatures_w1[-2:]
Middle_three_days = temperatures_w1[2:5]

print(First_three_days)
print(weekend)
print(Middle_three_days)
```

```python
# Step 5 : Create a 2D Array
Temperatures = np.array([
                         [22.5, 25.3, 20.8, 23.4, 26.1, 24.8, 21.9],     # week 1
                         [19.2, 22.5, 21.3, 24.0, 23.5, 22.8, 20.1]      # week 2
                        ])  
print("2D temperatures array:\n", Temperatures)
```

```python
# Step 6 : Inspect and Slice teh 2D Array
print("2D Shape:", Temperatures)
print("2D Data Type:", Temperatures)
print("2D Total Elements:", Temperatures)


# Extracting Individual weeks
Week_1 = Temperatures[0]
week_2 = Temperatures[1]

print(Week_1)
print(week_2)


# Extracting weekends (last two days) for both weeks
weekend_both = Temperatures[:,-2:]
print("weekends for both weeks:\n", weekend_both)
```

```python
import pandas as pd
```

```python
# Step 1: Creating pandas Series
Marks = pd.Series(
    data=[95, 92, 89, 85, 80], 
    index=['Rank1', 'Rank2', 'Rank3', 'Rank4', 'Rank5']
)
print("Marks Series:\n", Marks)
```

```python
# Step 2 : Indexing and Slicing
print("1st rank mark using integer position:", Marks.iloc[0])
print("3rd rank mark using iloc:", Marks.iloc[2])
print("Top 3 ranks using loc:\n", Marks.loc['Rank1':'rank3'])


# Boolean mask for ranks with marks greater than 90
high_scorers = Marks[Marks > 90]
print("ranks with marks > 90:\n", high_scorers)
```

```python
# Step 3 : Manipulating Series
Marks['Rank1'] = 100
print("After modifying 1st rank to 100:\n", Marks)

Marks = Marks.drop('Rank5')
print("After removing the last rank:\n", Marks)

CGPA = Marks / 10
print("Computed CGPA:\n", CGPA)
```

```python
import pandas as pd
```

```python
# Step 1 : Creating pandas DataFrame

transactions = pd.DataFrame({
                'TransactionID': [101, 102, 103, 104, 105, 106, 107, 108, 109, 110],
                'ProductionCategory': ['Electronics', 'Clothing', 'Electronics', 'Furniture', 'Clothing',
                                       'Electronics', 'Furniture', 'Clothing', 'Furniture', 'Electronics'],
                'Region': ['North', 'South', 'North', 'East', 'West', 'North', 
                           'East', 'West', 'South', 'North'],
                'Amount': [200, 150, 300, 450, 200, 250, 300, 180, 350, 400]})
print("Transactions DataFrame:\n", transactions)
```

```python
# Step 2 : Data Exploration

print("Head:\n", transactions.head())
print("Tail:\n", transactions.tail())
print("Shape:", transactions.shape)
print("Column Names:", transactions.columns.tolist())
print("Data types:\n", transactions.dtypes)
```

```python
# Display specific columns
print("Product Category and Amount:\n", transactions[['ProductionCategory','Amount']])
```

```python
# Retrieve the last 3 columns using iloc

print("Last 3 columns:\n", transactions.iloc[:, -3:])
```

```python
# Filter rows where Region is 'North' and amount is greater than 200

filtered_trans = transactions[(transactions['Region'] == 'North') & (transactions['Amount'] > 200)]
print("Filtered Transactions (North & Amount > 200):\n", filtered_trans)
```

```python
# Value counts and unique values

print("ProductCategory value counts:\n", transactions['ProductionCategory'].value_counts())
```

```python
# Group by 'region' and find the mean amount 

mean_amount_region = transactions.groupby('Region')['Amount'].mean()
print("mean_amount_region:\n", mean_amount_region)
```

```python
# Step 3 : Manipulating the DataFrame
# Modify Amount for TransactionID 102 to 165
transactions.loc[transactions['TransactionID'] == 102, 'Amount'] = 165

# Add a new column 'Discount' (10 % of Amount)
transactions['Discount'] = transactions['Amount'] * 0.10

# Remove the row with TransactionID 109
transactions = transactions[transactions['TransactionID'] != 109]

# Delete the 'Discount' column from the DataFrame
transactions = transactions.drop(columns=['Discount'])

print("Final Transactions  DataFrame:\n", transactions)
```
