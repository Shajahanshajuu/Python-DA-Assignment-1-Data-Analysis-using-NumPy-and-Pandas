```python
import numpy as np
```

```python
# Step 1 : Create 1D array for week 1 temparatures
temperatures_w1 = np.array([22.5, 25.3, 20.8, 23.4, 26.1, 24.8, 21.9])
print("temperatures_w1:", temperatures_w1)
```

```python
# Step 2 : Inspection and properties
print("Shape of temperatures_w1:", temperatures_w1.shape)
print("Data type of temperatures_w1:", temperatures_w1.dtype)
print("Number of elements in temperatures_w1:", temperatures_w1.size)
```

```python
# Step 3 : Array operations (Celsius to Fahrenheit)
fahrenheit_w1 = (temperatures_w1 * 9/5) + 32
print("fahrenheit_w1:", fahrenheit_w1)

# Statistical computations
max_temp = np.max(temperatures_w1)
min_temp = np.min(temperatures_w1)
mean_temp = np.mean(temperatures_w1)

print("Maximum Temperature (Celsius):", max_temp)
print("Minimum Temperature (Celsius):", min_temp)
print("Mean Temperature (Celsius):", mean_temp)
```

```python
# Step 4 : Array slicing and Indexing
First_three_days = temperatures_w1[:3]
weekend = temperatures_w1[-2:]
Middle_three_days = temperatures_w1[2:5]

print(First_three_days)
print(weekend)
print(Middle_three_days)
```

```python
# Step 5 : Create a 2D Array
Temperatures = np.array([
                         [22.5, 25.3, 20.8, 23.4, 26.1, 24.8, 21.9],     # week 1
                         [19.2, 22.5, 21.3, 24.0, 23.5, 22.8, 20.1]      # week 2
                        ])  
print("2D temperatures array:\n", Temperatures)
```

```python
# Step 6 : Inspect and Slice teh 2D Array
print("2D Shape:", Temperatures)
print("2D Data Type:", Temperatures)
print("2D Total Elements:", Temperatures)


# Extracting Individual weeks
Week_1 = Temperatures[0]
week_2 = Temperatures[1]

print(Week_1)
print(week_2)


# Extracting weekends (last two days) for both weeks
weekend_both = Temperatures[:,-2:]
print("weekends for both weeks:\n", weekend_both)
```
